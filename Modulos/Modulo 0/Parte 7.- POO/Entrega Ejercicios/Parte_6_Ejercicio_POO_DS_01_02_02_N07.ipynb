## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomesticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas ultimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas.
* Método que calcula la media de ventas de los ultimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
class Tienda:
    """
    Clase que representa una tienda de electrodomésticos y gestiona sus datos comerciales.

    Atributos de clase:
    ------------------
    tipo : str
        El tipo de negocio (Fijo: "Electrodomésticos").
    abierta : bool
        Estado de apertura de la tienda (Fijo: True).

    Atributos de instancia:
    -----------------------
    nombre : str
        Nombre comercial de la tienda.
    direccion : str
        Dirección física de la tienda.
    num_empleados : int
        Cantidad de empleados trabajando en la tienda.
    ventas_3_meses : list[int | float]
        Lista con los montos de ventas de los últimos 3 meses (de más antiguo a más reciente).
    """

    # Atributos comunes (de clase) con tipado explícito moderno
    tipo: str = "Electrodomésticos"
    abierta: bool = True

    def __init__(self, nombre: str, direccion: str, num_empleados: int, ventas_3_meses: list[int | float]) -> None:
        """Inicializa una nueva instancia de la clase Tienda."""
        self.nombre: str = nombre
        self.direccion: str = direccion
        self.num_empleados: int = num_empleados
        self.ventas_3_meses: list[int | float] = ventas_3_meses

    def total_ventas(self) -> int | float:
        """Calcula y devuelve la suma total de las ventas de los 3 meses."""
        return sum(self.ventas_3_meses)

    def media_ventas_por_empleado(self) -> int | float:
        """Calcula la media de ventas de los últimos meses por cada empleado."""
        if self.num_empleados == 0:
            return 0.0
        return self.total_ventas() / self.num_empleados

    def obtener_identificacion(self) -> str:
        """Devuelve un string formateado con el nombre de la tienda y su dirección."""
        return f"{self.nombre} - Ubicada en: {self.direccion}"

    def ventas_ultimo_mes(self) -> int | float:
        """Obtiene las ventas del último mes (el último elemento de la lista)."""
        return self.ventas_3_meses[-1] if self.ventas_3_meses else 0.0

    def proyeccion_marketing(self, inversion: int | float) -> list[int | float]:
        """
        Aplica una proyección de ventas según la inversión en marketing.
        
        Modifica el atributo ventas_3_meses y devuelve la nueva lista con el aumento.
        Si la inversión es menor a 1000, las ventas aumentan un 20% (x1.2).
        Si es mayor o igual a 1000, aumentan un 50% (x1.5).
        """
        factor: float = 1.5 if inversion >= 1000 else 1.2
        self.ventas_3_meses = [venta * factor for venta in self.ventas_3_meses]
        return self.ventas_3_meses


# ==========================================
# 1. Crear tres tiendas con datos inventados
# ==========================================
tienda1 = Tienda("Electro Outlet", "Calle Mayor 14", 5, [10000, 12000, 15000])
tienda2 = Tienda("TecnoHogar", "Avenida de la Constitución 45", 8, [20000, 25000, 22000])
tienda3 = Tienda("MegaElectro", "Avenida Gran Vía 102", 12, [35000, 40000, 38000])

todas_las_tiendas: list[Tienda] = [tienda1, tienda2, tienda3]


# ======================================================
# 2. Comprobar todo lo implementado en la Tienda 1
# ======================================================
print("=== COMPROBACIÓN DE LA TIENDA 1 ===")
print(f"Atributo común (Tipo): {tienda1.tipo}")
print(f"Atributo común (Abierta): {tienda1.abierta}")
print(f"Nombre: {tienda1.nombre}")
print(f"Dirección: {tienda1.direccion}")
print(f"Empleados: {tienda1.num_empleados}")
print(f"Ventas iniciales 3 meses: {tienda1.ventas_3_meses}")

# Métodos
print(f"Identificación: {tienda1.obtener_identificacion()}")
print(f"Total ventas: {tienda1.total_ventas()} €")
print(f"Media de ventas por empleado: {tienda1.media_ventas_por_empleado():.2f} €")
print(f"Ventas último mes: {tienda1.ventas_ultimo_mes()} €")

# Proyección de marketing (Inversión menor a 1000)
print(f"Proyección con 500€ de inversión en marketing: {tienda1.proyeccion_marketing(500)}")
print(f"Verificación del atributo modificado: {tienda1.ventas_3_meses}\n")


# ======================================================
# 3. Calcular las ventas del último mes de todas
# ======================================================
print("=== VENTAS DEL ÚLTIMO MES (TODAS LAS TIENDAS) ===")
for tienda in todas_las_tiendas:
    print(f"Tienda: {tienda.nombre} | Ventas último mes: {tienda.ventas_ultimo_mes()} €")
print()


# ======================================================
# 4. Tiendas cuya dirección incluye "Avenida"
# ======================================================
print("=== TIENDAS EN UNA AVENIDA ===")
for tienda in todas_las_tiendas:
    if "Avenida" in tienda.direccion:
        print(f"- {tienda.nombre} ({tienda.direccion})")
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    """
    Clase que representa a un perro, sus características físicas básicas
    y sus acciones principales como andar, parar o ladrar.

    Atributos de clase (comunes para todos los perros):
    --------------------------------------------------
    patas : int
        Número de patas del perro (Fijo: 4).
    orejas : int
        Número de orejas del perro (Fijo: 2).
    ojos : int
        Número de ojos del perro (Fijo: 2).

    Atributos de instancia (propios de cada perro):
    -----------------------------------------------
    raza : str
        La raza específica del perro.
    pelo : str
        El color del pelo del perro (Por defecto: "Marrón").
    dueno : str | None
        El nombre del dueño del perro (Por defecto: None / Sin dueño).
    velocidad : int | float
        La velocidad actual del perro (Inicialmente: 0).
    """

    # Atributos comunes seguros
    patas: int = 4
    orejas: int = 2
    ojos: int = 2

    def __init__(self, raza: str, pelo: str = "Marrón", dueno: str | None = None) -> None:
        """Inicializa un nuevo perro con su raza, pelo y dueño."""
        self.raza: str = raza
        self.pelo: str = pelo
        self.dueno: str | None = dueno
        self.velocidad: int | float = 0  # Por defecto, el perro está parado

    def andar(self, aumento_velocidad: int | float) -> None:
        """Suma el valor recibido a la velocidad actual del perro."""
        self.velocidad += aumento_velocidad

    def parar(self) -> None:
        """Detiene al perro restableciendo su velocidad a 0."""
        self.velocidad = 0

    def ladrar(self, argumento_entrada: str | int) -> str:
        """Devuelve un string combinando un ladrido con el texto recibido."""
        return "GUAU! " + str(argumento_entrada)


# ======================================================
# 1. Crear un objeto de tipo Perro, sin dueño
# ======================================================
mi_perro = Perro("Pastor Alemán")


# ======================================================
# 2. Comprobar que están bien todos sus atributos
# ======================================================
print("=== COMPROBACIÓN DE ATRIBUTOS DEL PERRO ===")
print(f"Patas: {mi_perro.patas}")
print(f"Orejas: {mi_perro.orejas}")
print(f"Ojos: {mi_perro.ojos}")
print(f"Raza: {mi_perro.raza}")
print(f"Color de pelo: {mi_perro.pelo}")
print(f"Dueño: {mi_perro.dueno} (Indica que no tiene)")
print(f"Velocidad inicial: {mi_perro.velocidad}\n")


# ======================================================
# 3. Prueba que ande y comprueba su velocidad
# ======================================================
print("=== PROBANDO EL MÉTODO ANDAR ===")
mi_perro.andar(5)
print(f"El perro empieza a caminar. Velocidad actual: {mi_perro.velocidad}")
mi_perro.andar(3.5)
print(f"El perro acelera. Nueva velocidad actual: {mi_perro.velocidad}\n")


# ======================================================
# 4. Páralo
# ======================================================
print("=== PROBANDO EL MÉTODO PARAR ===")
mi_perro.parar()
print(f"El perro se ha detenido. Velocidad actual: {mi_perro.velocidad}\n")


# Extra: Probar el método ladrar
print("=== PROBANDO EL MÉTODO LADRAR ===")
print(mi_perro.ladrar("Tengo hambre!"))
```
