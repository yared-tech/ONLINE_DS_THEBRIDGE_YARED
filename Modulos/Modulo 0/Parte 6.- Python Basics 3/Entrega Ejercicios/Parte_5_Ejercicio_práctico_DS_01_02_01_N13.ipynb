## Ejercicio 11
Convierte el programa del ejercicio 9 en una función al que se le pasen como argumentos la librería, posicional, y el título, argumento keyword con valor "ninguno", por defecto.

`````python
def buscar_libro(libreria, titulo="ninguno"):
    if titulo == "ninguno":
        return "Ese no lo tengo, ¿mola?"
    
    for libro in libreria:
        if libro["titulo"] == titulo:
            return libro
            
    return "Ese no lo tengo, ¿mola?"

print(buscar_libro(librería, titulo="1984"))
print(buscar_libro(librería, titulo="Harry Potter"))
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


`````python
def validar_dni(dni_completo):
    palabra = 'TRWAGMYFPDXBNJZSQVHLCKE'
    letra = dni_completo[-1].upper()
    numeros_texto = dni_completo[:-1]
    
    if not (7 <= len(numeros_texto) <= 8) or not numeros_texto.isdigit():
        return False
        
    dni_num = int(numeros_texto)
    return palabra[dni_num % 23] == letra

dni_usuario = input("Introduce tu DNI (con letra al final): ")
print(f"¿DNI válido?: {validar_dni(dni_usuario)}")
`````
